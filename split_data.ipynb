```python
# ----------------- مسیرها و پارامترها -----------------
data_path = os.path.join("dataset/Farsi_Type/")
train_labels_path = os.path.join(data_path, "train/")
test_labels_path = os.path.join(data_path, "test/")

train_images_path = os.path.join(data_path, "train", "images")
test_images_path = os.path.join(data_path, "test", "images")

concat_train_dir = "dataset/concatinated/train"
concat_test_dir = "dataset/concatinated/test"

NUM_TRAIN_SAMPLES = 20000
NUM_TEST_SAMPLES = 3500
# ----------------------------------------------------


def process_and_copy_split(labels_csv_path, src_img_dir, dst_dir, n_samples):
    dst_img_dir = os.path.join(dst_dir, "images")
    os.makedirs(dst_img_dir, exist_ok=True)

    df = pd.read_csv(labels_csv_path, header=None, encoding="utf-8-sig")
    df.columns = ["image_name", "text"]
    
    df["image_name"] = df["image_name"].astype(str).str.strip()

    existing_images = set(os.listdir(src_img_dir))
    df = df[df["image_name"].isin(existing_images)].reset_index(drop=True)

    sample_count = min(n_samples, len(df))
    sampled_df = df.sample(n=sample_count, random_state=42).reset_index(drop=True)

    print(f"در حال کپی {len(sampled_df)} تصویر به: {dst_img_dir}")

    for img_name in sampled_df["image_name"]:
        src = os.path.join(src_img_dir, img_name)
        dst = os.path.join(dst_img_dir, img_name)
        shutil.copy2(src, dst)

    out_csv_path = os.path.join(dst_dir, "labels.csv")
    sampled_df.to_csv(out_csv_path, header=False, index=False, encoding="utf-8-sig")
    print(f"فایل لیبل ذخیره شد در: {out_csv_path}\n")


print("--- پردازش Train ---")
process_and_copy_split(
    labels_csv_path=os.path.join(train_labels_path, "labels.csv"),
    src_img_dir=train_images_path,
    dst_dir=concat_train_dir,
    n_samples=NUM_TRAIN_SAMPLES
)

print("--- پردازش Test ---")
process_and_copy_split(
    labels_csv_path=os.path.join(test_labels_path, "labels.csv"),
    src_img_dir=test_images_path,
    dst_dir=concat_test_dir,
    n_samples=NUM_TEST_SAMPLES
)

print("عملیات با موفقیت انجام شد.")
```
